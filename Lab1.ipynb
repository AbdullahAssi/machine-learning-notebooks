```python
def primeNumber(n):
    if n==1:
        return False
    for i in range(2,n):
        if n%i==0:
            return False
    return True
num =  int(input("Enter a number: "))
if primeNumber(num):
    print(f"{num} is a Prime number")
else:
    print(f"{num} is not a Prime number")
```

```python
from collections import Counter

def is_palindrome(word):
    return word == word[::-1]

def count_palindromes(file_path):
    
    with open(file_path, 'r') as file:
        words = file.read().split()
    
    palindromes = [word for word in words if is_palindrome(word)]
    return Counter(palindromes)

file_path = 'text.txt' 
palindrome_counts = count_palindromes(file_path)
print(palindrome_counts)
```

```python
def is_disarium(num):
    # Converting the number to a string to easily access each digit
    num_str = str(num)
    length = len(num_str)
    
    # Initializing sum to 0
    sum_of_powers = 0
    
    # Creating a copy of the number
    temp = num
    
    # Calculating the sum of digits powered with their respective positions
    while temp > 0:
        single_digit = temp % 10
        sum_of_powers += single_digit ** length
        temp //= 10
        length -= 1
    
    # Checking if the sum of powers is equal to the original number
    return sum_of_powers == num

# Checking if the given number is a Disarium number
if is_disarium(num):
    print(f"{num} is a Disarium number")
else:
    print(f"{num} is not a Disarium number")
```
