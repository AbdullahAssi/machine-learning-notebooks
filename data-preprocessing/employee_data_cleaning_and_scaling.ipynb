# Employee Data Cleaning & Feature Scaling

Handling missing values and one-hot encoding categorical features on an employee dataset, then normalizing features with Min-Max and Standard scaling.

```python
import pandas as pd


df = pd.read_csv('employee_data.csv')

print(df.sample(5))
```

```python
df.isnull().sum()
```

```python
df = df.drop(index=7)
median_age = df['Age'].mean().astype(int)
df['Age'].fillna(median_age, inplace=True)

df.reset_index(drop=True, inplace=True)

df.to_csv('employee_data_processed.csv', index=False)

print(df.sample(5))
```

```python

median_salaries = df.groupby('Department')['Salary'].mean().astype(int)

def fill_missing_salary(row):
    if pd.isna(row['Salary']):
        return median_salaries[row['Department']]  
    return row['Salary']


df['Salary'] = df.apply(fill_missing_salary, axis=1)

df.to_csv('employee_data_processed.csv', index=False)

print(df)
```

```python

most_frequent_date = df['JoiningDate'].mode()[0]

df['JoiningDate'].fillna(most_frequent_date, inplace=True)

df.to_csv('employee_data_cleaned.csv', index=False)

print(df)
```

```python

duplicate_rows = df[df.duplicated()]


duplicate_columns = df.columns[df.columns.duplicated()]


if not duplicate_rows.empty:
    print("Duplicate Rows:")
    print(duplicate_rows)
else:
    print("No duplicate rows found.")

if len(duplicate_columns) > 0:
    print("Duplicate Columns:")
    print(duplicate_columns)
else:
    print("No duplicate columns found.")
```

```python

from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()

df[['Salary', 'PerformanceScore']] = scaler.fit_transform(df[['Salary', 'PerformanceScore']])

df.to_csv('employee_data_processed.csv', index=False)

print(df)
```

```python

from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()


df[['Salary', 'PerformanceScore']] = scaler.fit_transform(df[['Salary', 'PerformanceScore']])


df.to_csv('employee_data_processed.csv', index=False)

print(df)
```

```python

df_encoded = pd.get_dummies(df, columns=['Department'])


df_encoded.to_csv('employee_data_processed.csv', index=False)


print(df_encoded)
```

```python
from datetime import datetime


current_year = datetime.now().year


df['YearsAtCompany'] = current_year - pd.to_datetime(df['JoiningDate'], format='%d-%m-%y').dt.year


df.to_csv('employee_data_processed.csv', index=False)


print(df)
```
